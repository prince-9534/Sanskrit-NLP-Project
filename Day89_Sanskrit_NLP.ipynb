```python
import re
from collections import Counter

# 1. Sample Sanskrit Shloka (Bhagavad Gita 2.47)
sanskrit_text = """कर्मण्येवाधिकारस्ते मा फलेषु कदाचन।
मा कर्मफलहेतुर्भूर्मा ते सङ्गोऽस्त्वकर्मणि॥
"""
print("Original Text:")
print(sanskrit_text)

# 2. Tokenization (Words aur Punctuation alag karo)
# Regex use kar rahe hain (Devanagari ke liye)
words = re.findall(r'[\u0900-\u097F]+', sanskrit_text)
print(f"\nTotal Words: {len(words)}")
print(f"\nUnique Words: {len(set(words))}")
print(f"\nWords List: {words[:10]}...") # Pehle 10 words

# 3. Word Frequency (Kaunsa word kitni baar aaya)
word_counts = Counter(words)
print("\nTop 5 Most Frequent Words:")
for word, count in word_counts.most_common(5):
    print(f"{word}: {count} baar")
```

```python
# 1. Ek lamba Sanskrit sentence lete hain
sentence = "अहं संस्कृतम् पठामि। त्वं किं पठसि?"

# 2. Character-level analysis
print("Character Analysis")
for char in sentence[:20]: # Pehle 20 characters
    print(f"{char} -> Unicode: {hex(ord(char))}")

# 3. Words count
words = sentence.split() # Basic split
print(f"\nWords: {words}")
```

```python
# 1. English Sentence
eng_text = "I am learning NLP."
# 2. Sanskrit Sentence
sans_text = "अहं एन.एल.पी. पठामि।"

# Tokenize karo
eng_words = re.findall(r'\b\w+\b', eng_text.lower())
sans_words = re.findall(r'[\u0900-\u097F]+', sans_text)

print(f"English Words: {eng_words}")
print(f"Sanskrit Words: {sans_words}")

# Comparison
print(f"\nEnglish: {len(eng_words)} words")
print(f"Sanskrit: {len(sans_words)} words")
```
